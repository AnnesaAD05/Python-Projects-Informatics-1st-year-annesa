# ✈️ Travel Budget Planner


```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Mock API Data for demonstration purposes
# In a production environment, you could integrate real flight/hotel/currency APIs
DESTINATION_DATABASE = {
    "Tokyo": {
        "flight_per_person": 45000,    # in INR
        "hotel_per_night": 8000,       # in INR
        "food_per_day_person": 2500,   # in INR
        "transport_per_day_person": 800,# in INR
        "activities_fixed": 15000       # in INR
    },
    "Paris": {
        "flight_per_person": 55000,
        "hotel_per_night": 10000,
        "food_per_day_person": 3500,
        "transport_per_day_person": 1200,
        "activities_fixed": 20000
    },
    "New York": {
        "flight_per_person": 75000,
        "hotel_per_night": 12000,
        "food_per_day_person": 4000,
        "transport_per_day_person": 1500,
        "activities_fixed": 25000
    }
}

# Create interactive widgets
dest_widget = widgets.Dropdown(
    options=list(DESTINATION_DATABASE.keys()),
    value='Tokyo',
    description='Destination:',
)

duration_widget = widgets.IntSlider(
    value=7,
    min=1,
    max=30,
    step=1,
    description='Duration (Days):',
)

people_widget = widgets.IntSlider(
    value=2,
    min=1,
    max=10,
    step=1,
    description='People:',
)

calculate_btn = widgets.Button(
    description='Calculate Budget',
    button_style='success',
    icon='plane'
)

output_area = widgets.Output()

def calculate_budget(b):
    with output_area:
        clear_output()
        dest = dest_widget.value
        days = duration_widget.value
        people = people_widget.value

        data = DESTINATION_DATABASE[dest]

        # Estimation Logic
        flights = data["flight_per_person"] * people
        hotel = data["hotel_per_night"] * days * (1 if people <= 2 else (people // 2 + people % 2))
        food = data["food_per_day_person"] * days * people
        transport = data["transport_per_day_person"] * days * people
        activities = data["activities_fixed"] * people

        total = flights + hotel + food + transport + activities

        print(f"✈️ TRIP BUDGET FOR {dest.upper()}")
        print(f"Duration: {days} days | Travelers: {people}\n")
        print(f"Flights       ₹ {flights:,.2f}")
        print(f"Hotel         ₹ {hotel:,.2f}")
        print(f"Food          ₹ {food:,.2f}")
        print(f"Transport     ₹ {transport:,.2f}")
        print(f"Activities    ₹ {activities:,.2f}")
        print("─" * 40)
        print(f"Estimated Total: ₹ {total:,.2f}")

calculate_btn.on_click(calculate_budget)

# Display the widgets dashboard
display(widgets.VBox([dest_widget, duration_widget, people_widget, calculate_btn, output_area]))
```
