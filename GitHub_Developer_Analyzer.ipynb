## GitHub Developer Analyzer



```python
import requests
from collections import Counter

def analyze_github_user(username):
    # Base API URLs
    user_url = f"https://api.github.com/users/{username}"
    repos_url = f"https://api.github.com/users/{username}/repos?per_page=100"

    # Fetch basic user info
    user_response = requests.get(user_url)
    if user_response.status_code != 200:
        print(f"Error: Unable to fetch data for user '{username}'. Status code: {user_response.status_code}")
        return

    user_data = user_response.json()

    # Fetch repositories
    repos_response = requests.get(repos_url)
    repos_data = repos_response.json() if repos_response.status_code == 200 else []

    # Aggregate statistics
    public_repos = user_data.get('public_repos', 0)
    followers = user_data.get('followers', 0)
    following = user_data.get('following', 0)

    total_stars = 0
    total_forks = 0
    languages = []

    for repo in repos_data:
        total_stars += repo.get('stargazers_count', 0)
        total_forks += repo.get('forks_count', 0)
        lang = repo.get('language')
        if lang:
            languages.append(lang)

    # Calculate language percentages for top languages
    lang_counts = Counter(languages)
    total_lang_repos = sum(lang_counts.values())

    # Print ASCII Card
    print("╔═══════════════════════════════════════════╗")
    print("║             DEVELOPER PROFILE             ║")
    print("╚═══════════════════════════════════════════╝")
    print(f" User:             {user_data.get('login', username)}")
    print(f" Name:             {user_data.get('name', 'N/A')}")
    print(f" Repositories:     {public_repos}")
    print(f" Followers:        {followers}")
    print(f" Following:        {following}")
    print(f" Total Stars:      {total_stars}")
    print(f" Total Forks:      {total_forks}")
    print("\n Top Languages:")

    if total_lang_repos > 0:
        for lang, count in lang_counts.most_common(5):
            percentage = count / total_lang_repos
            bar_length = int(percentage * 20)
            bar = "█" * bar_length
            print(f"  {lang:<16} {bar}")
    else:
        print("  No language data available.")
    print("═════════════════════════════════════════════")
```

```python
# Example usage:
analyze_github_user('AnnesaAD05')
```
