# 💰 Expense-to-Exchange Converter


```python
import re
import requests

def get_live_rates():
    """Fetches live exchange rates from a free API with USD as the base."""
    try:
        response = requests.get("https://open.er-api.com/v6/latest/USD")
        if response.status_status == 200 or response.ok:
            data = response.json()
            rates = data.get("rates", {})
            return {
                "INR": rates.get("INR", 83.5),
                "EUR": rates.get("EUR", 0.92),
                "GBP": rates.get("GBP", 0.78),
                "JPY": rates.get("JPY", 155.0)
            }
    except Exception:
        pass
    # Fallback rates in case API is offline
    return {"INR": 83.5, "EUR": 0.92, "GBP": 0.78, "JPY": 155.0}

def parse_and_convert_expenses(expense_text):
    # Extract all numbers preceded by optionally a dollar sign
    amounts = re.findall(r'\$?\s*(\d+(?:\.\d{2})?)', expense_text)
    total_usd = sum(float(amount) for amount in amounts)

    rates = get_live_rates()

    print("--- EXPENSE SUMMARY ---")
    print(f"TOTAL USD: ${total_usd:,.2f}\n")
    print("CONVERSIONS:")
    print(f"≈ ₹{total_usd * rates['INR']:,.2f} (INR)")
    print(f"≈ €{total_usd * rates['EUR']:,.2f} (EUR)")
    print(f"≈ £{total_usd * rates['GBP']:,.2f} (GBP)")
    print(f"≈ ¥{total_usd * rates['JPY']:,.2f} (JPY)")
    return total_usd

# Test input matching the prompt
user_input = """
Hotel: $120
Food: $40
Transport: $25
"""
parse_and_convert_expenses(user_input)
```

## ✈️ Country-to-Country Travel Budget Translator
This feature allows users to enter a pair of countries (e.g., `India → Japan`) to estimate an equivalent daily travel budget based on Purchasing Power Parity (PPP) approximations.

```python
def travel_budget_translator(route_str, base_budget=10000):
    # Simplified cost relative index mapping (Base USD index = 1.0)
    cost_indices = {
        "india": 0.35,
        "japan": 0.75,
        "usa": 1.00,
        "germany": 0.80,
        "uk": 0.85
    }

    parts = [p.strip().lower() for p in route_str.split("→")]
    if len(parts) != 2:
        print("Please write the route in the format: Country A → Country B")
        return

    source, destination = parts[0], parts[1]

    if source in cost_indices and destination in cost_indices:
        ratio = cost_indices[destination] / cost_indices[source]
        equivalent_budget = base_budget * ratio
        print(f"For a base budget of {base_budget:,.2f} in {source.title()},")
        print(f"you will need approximately {equivalent_budget:,.2f} equivalent buying power in {destination.title()}.")
    else:
        print(f"Supported countries for simulation: {', '.join(cost_indices.keys()).title()}")

travel_budget_translator("India → Japan", base_budget=50000)
```
