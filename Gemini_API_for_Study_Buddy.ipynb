```python
if "[PASTE YOUR LONG PARAGRAPH / LECTURE NOTES HERE]" not in lecture_notes:
    prompt = f"""Analyze the following text and provide

SUMMARY
────────────
(A concise summary of the key points)

IMPORTANT TERMS
────────────
(List 5-10 important terms with brief definitions or explanations)

5 MARK QUESTIONS# The original placeholder check is no longer needed since the notes are directly provided.
# The 'if' block below is removed because the notes are now directly assigned.
# if "[PASTE YOUR LONG PARAGRAPH / LECTURE NOTES HERE]" in lecture_notes:
#     print("Please replace the placeholder text in the 'lecture_notes' variable with your actual content.")
# else:
────────────
(Generate 2-3 detailed questions that would typically be worth 5 marks, requiring comprehensive answers)

MCQs
────────────
(Generate 3-5 Multiple Choice Questions with 4 options (A, B, C, D) and indicate the correct answer. Format as Q1. Question\nA. Option 1\nB. Option 2\nC. Option 3\nD. Option 4\nCorrect Answer: X)

TEXT:
{lecture_notes}
"""

    print("Generating study material...")

    model = None
    preferred_models = ['gemini-3.8-flash', 'gemini-pro'] # Prioritize newer models
    available_models = {m.name: m for m in genai.list_models()}

    for p_model_name in preferred_models:
        full_model_name = f'models/{p_model_name}'
        if full_model_name in available_models and 'generateContent' in available_models[full_model_name].supported_generation_methods:
            model = genai.GenerativeModel(full_model_name)
            print(f"Using preferred model: {full_model_name}")
            break

    # Fallback to any other suitable model if preferred models are not found
    if model is None:
        for m in genai.list_models():
            if 'generateContent' in m.supported_generation_methods:
                model = genai.GenerativeModel(m.name)
                print(f"Using fallback model: {m.name}")
                break

    if model is None:
        print("No suitable generative model found. Please check your API key and model availability.")
    else:
        # Call generate_content on the model instance
        response = model.generate_content(contents=prompt)

        # Display the generated content
        print(response.text)
else:
    print("Please provide your lecture notes in the previous cell to generate content.")
```
