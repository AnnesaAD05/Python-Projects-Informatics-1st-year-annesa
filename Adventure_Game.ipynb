# Text-Based Adventure Game

```python
rooms = {
    'start_room': {
        'description': 'You are in a dimly lit antechamber. There is a wooden door to the north and a dusty corridor to the east.',
        'exits': {'north': 'wooden_door_room', 'east': 'corridor'},
        'items': ['rusty key']
    },
    'wooden_door_room': {
        'description': 'You are in a small, empty room. There is a heavy wooden door to the south.',
        'exits': {'south': 'start_room'},
        'items': []
    },
    'corridor': {
        'description': 'You are in a long, dark corridor. There is a faint light coming from the west, and a mysterious opening to the south.',
        'exits': {'west': 'start_room', 'south': 'mysterious_cave'},
        'items': ['torch']
    },
    'mysterious_cave': {
        'description': "You have stumbled into a mysterious cave. It's too dark to see much without a light source. There is a narrow passage back to the north.",
        'exits': {'north': 'corridor'},
        'items': []
    }
}

items = ['rusty key', 'torch', 'sword', 'shield']
```

```python
current_room = 'start_room'
inventory = []
```

```python
def display_room():
    global current_room
    room_data = rooms[current_room]
    print(f"\n--- {current_room.replace('_', ' ').title()} ---")
    print(room_data['description'])

    if room_data['items']:
        print("You see: " + ', '.join(room_data['items']))

    exits = ', '.join(room_data['exits'].keys())
    print(f"Exits: {exits}")

def main_game_loop():
    global current_room, inventory
    game_over = False

    while not game_over:
        display_room()
        command = input("\nWhat do you want to do? ").lower().strip()

        if command == 'quit':
            game_over = True
            print("Thanks for playing!")
        elif command.startswith('go '):
            direction = command[3:].strip()
            if direction in rooms[current_room]['exits']:
                current_room = rooms[current_room]['exits'][direction]
            else:
                print("You can't go that way.")
        elif command.startswith('take '):
            item_to_take = command[5:].strip()
            if item_to_take in rooms[current_room]['items']:
                inventory.append(item_to_take)
                rooms[current_room]['items'].remove(item_to_take)
                print(f"You picked up the {item_to_take}.")
            else:
                print(f"There is no {item_to_take} here.")
        elif command == 'inventory' or command == 'i':
            if inventory:
                print("Your inventory: " + ', '.join(inventory))
            else:
                print("Your inventory is empty.")
        else:
            print("I don't understand that command. Try 'go [direction]', 'take [item]', 'inventory', or 'quit'.")

# Uncomment the line below to start the game:
main_game_loop()
```
