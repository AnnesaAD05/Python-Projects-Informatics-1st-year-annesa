```python
from pathlib import Path
import re # Import regex for pattern matching

def visualize_directory_tree(path, indent="", is_last=False, max_depth=None, current_depth=0, exclude_patterns=None):
    """
    Recursively visualizes the directory tree from the given path.

    Args:
        path (Path): The starting path (directory or file).
        indent (str): The current indentation string for tree branches.
        is_last (bool): True if the current item is the last sibling in its parent directory.
        max_depth (int, optional): Maximum depth to visualize. None for no limit.
        current_depth (int): Current recursion depth. Used internally.
        exclude_patterns (list, optional): A list of regex patterns for files/directories to exclude.
    """
    if not path.exists():
        print(f"Error: Path '{path}' does not exist.")
        return


    if exclude_patterns:
        for pattern in exclude_patterns:
            if re.search(pattern, path.name):
                return # Exclude this path


    if max_depth is not None and current_depth > max_depth:
        return


    prefix = "└── " if is_last else "├── "
    print(f"{indent}{prefix}{path.name}/")

    if path.is_dir():

        next_indent = indent + ("    " if is_last else "│   ")

        children = []
        for child in sorted(list(path.iterdir())):

            if exclude_patterns:
                excluded = False
                for pattern in exclude_patterns:
                    if re.search(pattern, child.name):
                        excluded = True
                        break
                if excluded:
                    continue
            children.append(child)

        for i, child in enumerate(children):
            is_last_child = (i == len(children) - 1)
            if child.is_dir():
                visualize_directory_tree(child, next_indent, is_last_child, max_depth, current_depth + 1, exclude_patterns)
            else:
                file_prefix = "└── " if is_last_child else "├── "
                print(f"{next_indent}{file_prefix}{child.name}")
```

```python
import os
import shutil

# Define a temporary directory for demonstration
tmp_dir = Path("temp_tree_demo")

# --- Directory setup for this demonstration ---
# Clean up previous demo if it exists
if tmp_dir.exists():
    shutil.rmtree(tmp_dir)

# Create the directory structure
tmp_dir.mkdir()
(tmp_dir / "file1.txt").touch()
(tmp_dir / "subdir1").mkdir()
(tmp_dir / "subdir1" / "file2.md").touch()
(tmp_dir / "subdir1" / "subdir2").mkdir()
(tmp_dir / "subdir1" / "subdir2" / "file3.py").touch()
(tmp_dir / "subdir1" / "subdir2" / "empty_dir").mkdir()
(tmp_dir / "file4.jpg").touch()
# ------------------------------------------------

print("\n--- Visualizing with max_depth=1 ---")
visualize_directory_tree(tmp_dir, max_depth=1)

print("\n--- Visualizing excluding files ending with .txt ---")
visualize_directory_tree(tmp_dir, exclude_patterns=[r'\.txt$'])

print("\n--- Visualizing excluding 'subdir1' and files ending with .py ---")
visualize_directory_tree(tmp_dir, exclude_patterns=['subdir1', r'\.py$'])

print("\n--- Visualizing with max_depth=2 and excluding 'empty_dir' ---")
visualize_directory_tree(tmp_dir, max_depth=2, exclude_patterns=['empty_dir'])

# Clean up the temporary directory after all demonstrations
shutil.rmtree(tmp_dir)
print(f"\nCleaned up temporary directory: {tmp_dir}")
```
