```python
from geopy.geocoders import Nominatim
from geopy.distance import geodesic
from gtts import gTTS
import ollama
import time
import gradio as gr # Added this import

geolocator = Nominatim(user_agent="delhi_tour_guide")

def get_coordinates(location_name):
    try:
        location = geolocator.geocode(location_name + ", Delhi")
        if location: return (location.latitude, location.longitude)
        return None
    except: return None

def find_nearest_landmark(user_lat, user_lon, radius_km=5.0):
    # Retrieve all from Chroma (for proximity check)
    results = collection.get(include=["metadatas", "documents"])
    nearby = []

    for i in range(len(results['ids'])):
        meta = results['metadatas'][i]
        dist = geodesic((user_lat, user_lon), (meta['lat'], meta['lon'])).km
        if dist <= radius_km:
            nearby.append({"name": meta['name'], "dist": dist, "doc": results['documents'][i], "meta": meta}) # Added "meta": meta

    return sorted(nearby, key=lambda x: x['dist'])[0] if nearby else None

def generate_tour_guide_narration(landmark_name, context):
    prompt = f"Landmark: {landmark_name}\nContext: {context}"
    system_msg = "You are an expert, engaging Delhi historical tour guide. Using ONLY the provided context, write a vibrant, architectural, and historical narration suitable for an audio tour. Keep it concise but captivating."

    max_retries = 3
    for attempt in range(max_retries):
        try:
            response = ollama.chat(model='llama3.2', messages=[
                {'role': 'system', 'content': system_msg},
                {'role': 'user', 'content': prompt},
            ])
            return response['message']['content']
        except ConnectionError as e:
            if attempt < max_retries - 1:
                print(f"Ollama Connection Error (attempt {attempt + 1}/{max_retries}): {e}. Retrying in 5 seconds...")
                time.sleep(5)
            else:
                print(f"Ollama Connection Error (final attempt): {e}. "
                      "Please ensure the Ollama server is running. You might need to re-run the cell that starts Ollama (cell 01aa63d4).")
                raise # Re-raise the exception after all retries fail

def text_to_speech(text, filename="narration.mp3"):
    tts = gTTS(text=text, lang='en')
    tts.save(filename)
    return filename

# The content moved from a194a7f6 starts here
def tour_app(street_name, lat, lon, radius):
    # 1. Determine Coordinates
    current_lat, current_lon = (lat, lon)
    if street_name:
        coords = get_coordinates(street_name)
        if coords:
            current_lat, current_lon = coords

    # 2. Find Landmark
    landmark = find_nearest_landmark(current_lat, current_lon, radius)
    if not landmark:
        return "No landmarks found in this radius.", None, "Try increasing the search radius.", "N/A", "N/A"

    # Extract metro information
    metro_station = landmark['meta'].get('nearest_metro_station', 'N/A')
    metro_line = landmark['meta'].get('metro_line', 'N/A')

    # 3. Generate Narration via Llama 3.2
    narration = generate_tour_guide_narration(landmark['name'], landmark['doc'])

    # 4. Generate Audio
    audio_path = text_to_speech(narration)

    return landmark['name'], audio_path, narration, metro_station, metro_line

# UI Construction
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🕌 Delhi Historical Landmark Tour Guide")
    gr.Markdown("Enter a location in Delhi to discover nearby history via AI-powered narration, including nearest Metro routes.")

    with gr.Row():
        with gr.Column():
            street_input = gr.Textbox(label="Street or Location Name (e.g., 'Connaught Place')")
            lat_input = gr.Number(label="Latitude", value=28.6139)
            lon_input = gr.Number(label="Longitude", value=77.2090)
            radius_slider = gr.Slider(minimum=1, maximum=20, value=5, label="Search Radius (km)")
            search_btn = gr.Button("Start Tour", variant="primary")

        with gr.Column():
            landmark_output = gr.Textbox(label="Identified Landmark")
            metro_station_output = gr.Textbox(label="Nearest Metro Station")
            metro_line_output = gr.Textbox(label="Metro Line")
            audio_output = gr.Audio(label="Audio Narration")
            text_output = gr.Textbox(label="Full Narration Text", lines=10)

    search_btn.click(fn=tour_app,
                     inputs=[street_input, lat_input, lon_input, radius_slider],
                     outputs=[landmark_output, audio_output, text_output, metro_station_output, metro_line_output])

demo.launch(share=True, debug=True)


unknown_monument_name, unknown_monument_audio, unknown_monument_narration, unknown_metro_station, unknown_metro_line = tour_app(
    street_name='Imaginary Monument of Delhi',
    lat=28.6139, # Using a central Delhi coordinate for context
    lon=77.2090,
    radius=1 # Small radius to ensure it's not accidentally near a known landmark
)

print(f"Landmark Name: {unknown_monument_name}")
print(f"Narration: {unknown_monument_narration}")
print(f"Nearest Metro Station: {unknown_metro_station}")
print(f"Metro Line: {unknown_metro_line}")
```
