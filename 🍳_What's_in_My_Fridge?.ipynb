```python
import requests
from google.colab import userdata

# Retrieve API key from Colab secrets
SPOONACULAR_API_KEY = userdata.get('SPOONACULAR_API_KEY')

def get_recipes_by_ingredients(ingredients, number=3, ranking=1):
    """
    Fetches recipes based on a list of ingredients using the Spoonacular API.

    Args:
        ingredients (list): A list of ingredient strings.
        number (int): The maximum number of recipes to return (default is 3).
        ranking (int): Whether to maximize used ingredients (1) or minimize missing ingredients (2).
                       (default is 1, maximize used ingredients).

    Returns:
        list: A list of dictionaries, where each dictionary represents a recipe.
    """
    base_url = "https://api.spoonacular.com/recipes/findByIngredients"
    ingredients_str = ",".join(ingredients)

    params = {
        "ingredients": ingredients_str,
        "number": number,
        "ranking": ranking, # 1: maximize used ingredients, 2: minimize missing ingredients
        "apiKey": SPOONACULAR_API_KEY
    }

    try:
        response = requests.get(base_url, params=params)
        response.raise_for_status()  # Raise an HTTPError for bad responses (4xx or 5xx)
        return response.json()
    except requests.exceptions.HTTPError as http_err:
        print(f"HTTP error occurred: {http_err}")
        if response.status_code == 401:
            print("Please check your Spoonacular API key.")
        elif response.status_code == 402:
            print("Your daily points limit of 150 has been reached. Please check the Spoonacular dashboard.")
        return []
    except Exception as err:
        print(f"An error occurred: {err}")
        return []
```

```python
input_ingredients = ['potato', 'onion', 'tomato', 'egg']
recipes = get_recipes_by_ingredients(input_ingredients)

if recipes:
    print(f"🍳 Recipes you can make with {', '.join(input_ingredients)}:")
    for i, recipe in enumerate(recipes):
        print(f"{i+1}. {recipe['title']}")
        if recipe.get('usedIngredients'):
            used = ', '.join([ing['name'] for ing in recipe['usedIngredients']])
            print(f"   - Used ingredients: {used}")
        if recipe.get('missedIngredients'):
            missed = ', '.join([ing['name'] for ing in recipe['missedIngredients']])
            print(f"   - Missing ingredients: {missed}")
else:
    print("No recipes found or an error occurred. Please check your ingredients and API key.")
```
