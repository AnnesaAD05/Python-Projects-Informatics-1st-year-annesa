```python
# Define your flashcards
flashcards = {
    "Python": "A high-level, interpreted programming language known for its readability and versatility.",
    "Variable": "A named storage location in memory used to store data.",
    "Function": "A block of organized, reusable code that performs a single, related action.",
    "Loop": "A control flow statement that allows code to be executed repeatedly.",
    "List": "An ordered, mutable collection of items in Python."
}

print(f"Loaded {len(flashcards)} flashcards.")
```

```python
\import random


score = 0
incorrect_answers = {}

card_list = list(flashcards.items())
random.shuffle(card_list)

print("\n--- Flashcard Quiz Start ---\n")

for term, correct_definition in card_list:
    print(f"Term: {term}")
    user_answer = input("Your definition: ")


    if user_answer.strip().lower() == correct_definition.strip().lower():
        print("Correct!\n")
        score += 1
    else:
        print(f"Incorrect. The correct definition was: {correct_definition}\n")
        incorrect_answers[term] = {'your_answer': user_answer, 'correct_answer': correct_definition}

print("\n--- Quiz End ---\n")
print(f"You scored {score} out of {len(flashcards)} correct!\n")

if incorrect_answers:
    print("--- Review Incorrect Answers ---")
    for term, answers in incorrect_answers.items():
        print(f"Term: {term}")
        print(f"  Your answer: {answers['your_answer']}")
        print(f"  Correct answer: {answers['correct_answer']}\n")
else:
    print("Great job! You got all of them correct!")
```
