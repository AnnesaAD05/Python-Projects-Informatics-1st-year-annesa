```python
import pyfiglet

# Get user input for their name
user_name = input("Please enter your name: ")
```

```python
# Generate ASCII art from the name
ascii_art_name = pyfiglet.figlet_format(user_name)

# Print the ASCII art
print(ascii_art_name)
```

```python
print("Available fonts (showing first 10 for brevity):\n")
all_fonts = pyfiglet.FigletFont.getFonts()

for i, font_name in enumerate(all_fonts[:10]):
    print(f"{i+1}. {font_name}")

if len(all_fonts) > 10:
    print(f"... and {len(all_fonts) - 10} more fonts.")

selected_font = input("\nEnter the name of a font you'd like to use (e.g., 'slant', 'chunky', 'isometric1'): ")

if selected_font in all_fonts:
    custom_fig = pyfiglet.Figlet(font=selected_font)
    ascii_art_custom = custom_fig.renderText(user_name)
    print(f"\nYour name '{user_name}' in '{selected_font}' font:\n")
    print(ascii_art_custom)
else:
    print(f"Error: Font '{selected_font}' not found. Please choose from the available fonts.")
```
