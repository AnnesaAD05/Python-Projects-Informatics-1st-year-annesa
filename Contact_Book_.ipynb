```python
class Contact:
    def __init__(self, name, phone_number):
        self.name = name
        self.phone_number = phone_number

    def __str__(self):
        return f"Name: {self.name}, Phone: {self.phone_number}"

    def __repr__(self):
        return f"Contact('{self.name}', '{self.phone_number}')"
```

```python
class ContactBook:
    def __init__(self):
        self.contacts = []

    def add_contact(self, name, phone_number):
        contact = Contact(name, phone_number)
        self.contacts.append(contact)
        print(f"Contact '{name}' added.")

    def search_contact(self, name):
        found_contacts = [contact for contact in self.contacts if name.lower() in contact.name.lower()]
        if found_contacts:
            print(f"Found {len(found_contacts)} contact(s) matching '{name}':")
            for contact in found_contacts:
                print(contact)
            return found_contacts
        else:
            print(f"No contact found with name '{name}'.")
            return []

    def update_contact(self, old_name, new_name=None, new_phone_number=None):
        found = False
        for contact in self.contacts:
            if contact.name.lower() == old_name.lower():
                if new_name:
                    contact.name = new_name
                if new_phone_number:
                    contact.phone_number = new_phone_number
                print(f"Contact '{old_name}' updated.")
                found = True
                break
        if not found:
            print(f"Contact '{old_name}' not found.")

    def delete_contact(self, name):
        original_count = len(self.contacts)
        self.contacts = [contact for contact in self.contacts if contact.name.lower() != name.lower()]
        if len(self.contacts) < original_count:
            print(f"Contact '{name}' deleted.")
        else:
            print(f"Contact '{name}' not found.")

    def list_all_contacts(self):
        if self.contacts:
            print("All Contacts:")
            for contact in self.contacts:
                print(contact)
        else:
            print("Contact book is empty.")
```

```python
my_contact_book = ContactBook()

my_contact_book.add_contact("Alice", "111-222-3333")
my_contact_book.add_contact("Bob", "444-555-6666")
my_contact_book.add_contact("Charlie", "777-888-9999")
my_contact_book.add_contact("Alicia", "123-456-7890")

print("\n--- After adding contacts ---")

my_contact_book.list_all_contacts()

print("\n--- Searching for 'Alice' ---")

my_contact_book.search_contact("Alice")

print("\n--- Searching for 'David' (not found) ---")
my_contact_book.search_contact("David")

print("\n--- Updating 'Bob' ---")

my_contact_book.update_contact("Bob", new_phone_number="000-111-2222")
my_contact_book.update_contact("Charlie", new_name="Charles", new_phone_number="101-202-3030")

print("\n--- After updating contacts ---")
my_contact_book.list_all_contacts()

print("\n--- Deleting 'Alice' ---")

my_contact_book.delete_contact("Alice")

print("\n--- Deleting 'Frank' (not found) ---")
my_contact_book.delete_contact("Frank")

print("\n--- Final list of contacts ---")

my_contact_book.list_all_contacts()
```
