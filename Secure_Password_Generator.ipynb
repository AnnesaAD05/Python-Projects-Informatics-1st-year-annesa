```python
# Import the necessary libraries
import random
import string

print('Libraries imported successfully.')
```

```python
def generate_password(length, include_numbers=True, include_symbols=True):
    """
    Generates a secure, randomized password.

    Args:
        length (int): The desired length of the password.
        include_numbers (bool): Whether to include numbers in the password.
        include_symbols (bool): Whether to include symbols in the password.

    Returns:
        str: The generated password.
    """
    characters = string.ascii_letters
    if include_numbers:
        characters += string.digits
    if include_symbols:
        characters += string.punctuation

    if not characters:
        raise ValueError("No character types selected for password generation.")

    password = ''.join(random.choice(characters) for i in range(length))
    return password

print('Password generation function defined.')
```

```python

# Handle user input and generate password
def get_user_input():
    while True:
        try:
            length = int(input("Enter desired password length (e.g., 12): "))
            if length <= 0:
                print("Password length must be a positive number.")
            else:
                break
        except ValueError:
            print("Invalid input. Please enter a number for length.")

    while True:
        include_numbers_str = input("Include numbers? (yes/no): ").lower()
        if include_numbers_str in ['yes', 'no']:
            include_numbers = (include_numbers_str == 'yes')
            break
        else:
            print("Invalid input. Please type 'yes' or 'no'.")

    while True:
        include_symbols_str = input("Include symbols? (yes/no): ").lower()
        if include_symbols_str in ['yes', 'no']:
            include_symbols = (include_symbols_str == 'yes')
            break
        else:
            print("Invalid input. Please type 'yes' or 'no'.")

    return length, include_numbers, include_symbols

# Get user preferences
password_length, include_nums, include_syms = get_user_input()

# Generate the password
try:
    secure_password = generate_password(password_length, include_nums, include_syms)
    print(f"\nGenerated Password: {secure_password}")
except ValueError as e:
    print(f"Error: {e}")
```

```python
from cryptography.fernet import Fernet
import os

def generate_key():
    """
    Generates a new Fernet encryption key.
    """
    return Fernet.generate_key()

def save_key(key, filename="secret.key"):
    """
    Saves the encryption key to a file.
    It's crucial to keep this key secure.
    """
    with open(filename, "wb") as key_file:
        key_file.write(key)
    print(f"Encryption key saved to {filename}")

def load_key(filename="secret.key"):
    """
    Loads the encryption key from a file.
    """
    if not os.path.exists(filename):
        print(f"Key file '{filename}' not found. Please ensure the key exists or generate a new one.")
        return None
    with open(filename, "rb") as key_file:
        key = key_file.read()
    return key

# Generate a key if it doesn't exist, or load it
key_filename = "secret.key"
if not os.path.exists(key_filename):
    encryption_key = generate_key()
    save_key(encryption_key, key_filename)
else:
    encryption_key = load_key(key_filename)
    print(f"Encryption key loaded from {key_filename}")

# Initialize the Fernet cipher with the key
if encryption_key:
    fernet = Fernet(encryption_key)
    print("Fernet cipher initialized.")
else:
    print("Could not initialize Fernet cipher: no key available.")
```

```python
def encrypt_password(password, fernet_cipher):
    """
    Encrypts a password using the provided Fernet cipher.
    """
    return fernet_cipher.encrypt(password.encode())

def decrypt_password(encrypted_password, fernet_cipher):
    """
    Decrypts an encrypted password using the provided Fernet cipher.
    """
    return fernet_cipher.decrypt(encrypted_password).decode()

def save_password_securely(password, filename="passwords.txt", fernet_cipher=None):
    """
    Encrypts and saves a password to a specified file.
    """
    if not fernet_cipher:
        print("Error: Fernet cipher not initialized. Cannot save securely.")
        return

    encrypted_password = encrypt_password(password, fernet_cipher)
    with open(filename, "ab") as f:
        f.write(encrypted_password + b'\n') # Add newline for multiple passwords
    print(f"Password securely saved to {filename}")

def load_passwords_securely(filename="passwords.txt", fernet_cipher=None):
    """
    Loads and decrypts all passwords from a specified file.
    """
    if not fernet_cipher:
        print("Error: Fernet cipher not initialized. Cannot load securely.")
        return []

    if not os.path.exists(filename):
        print(f"No password file found at {filename}")
        return []

    decrypted_passwords = []
    with open(filename, "rb") as f:
        for line in f:
            try:
                encrypted_password = line.strip()
                if encrypted_password:
                    decrypted_passwords.append(decrypt_password(encrypted_password, fernet_cipher))
            except Exception as e:
                print(f"Could not decrypt a password line. It might be corrupted or encrypted with a different key: {e}")
    return decrypted_passwords

print('Encryption/decryption and secure save/load functions defined.')

# Example usage:
if 'secure_password' in locals() and fernet:
    save_password_securely(secure_password, fernet_cipher=fernet)
    print("\n--- Stored Passwords ---")
    loaded_passwords = load_passwords_securely(fernet_cipher=fernet)
    for i, pwd in enumerate(loaded_passwords):
        print(f"Password {i+1}: {pwd}")
else:
    print("No password generated yet or Fernet cipher not initialized. Skipping secure save/load example.")
```
