```python
conversion_rates = {
    'length': {
        'meters': 1.0,
        'kilometers': 1000.0,
        'miles': 1609.34,
        'yards': 0.9144,
        'feet': 0.3048,
        'inches': 0.0254
    },
    'weight': {
        'kilograms': 1.0,
        'grams': 0.001,
        'pounds': 0.453592,
        'ounces': 0.0283495
    },
    'temperature': {
        'celsius': {'to_kelvin': lambda c: c + 273.15, 'from_kelvin': lambda k: k - 273.15},
        'fahrenheit': {'to_kelvin': lambda f: (f - 32) * 5/9 + 273.15, 'from_kelvin': lambda k: (k - 273.15) * 9/5 + 32},
        'kelvin': {'to_kelvin': lambda k: k, 'from_kelvin': lambda k: k} # Kelvin is the base for temperature conversion
    }
}

print("Conversion rates defined.")
```

```python
def display_main_menu():
    print("\n--- Universal Unit Converter ---")
    print("Select a category for conversion:")
    for i, category in enumerate(conversion_rates.keys(), 1):
        print(f"{i}. {category.capitalize()}")
    print("0. Exit")

    while True:
        try:
            choice = int(input("Enter your choice (number): "))
            if 0 <= choice <= len(conversion_rates):
                return choice
            else:
                print("Invalid choice. Please enter a number from the menu.")
        except ValueError:
            print("Invalid input. Please enter a number.")

print("Main menu function defined.")
```

## Conversion Logic for Each Category

```python
def get_unit_input(category_name, available_units):
    print(f"\nAvailable units for {category_name.capitalize()}:")
    for unit in available_units:
        print(f"- {unit}")

    while True:
        from_unit = input(f"Enter the unit to convert FROM (e.g., {available_units[0]}): ").lower()
        if from_unit in available_units:
            break
        else:
            print("Invalid unit. Please choose from the available units.")

    while True:
        to_unit = input(f"Enter the unit to convert TO (e.g., {available_units[1]}): ").lower()
        if to_unit in available_units:
            break
        else:
            print("Invalid unit. Please choose from the available units.")
    return from_unit, to_unit

def convert_length(value, from_unit, to_unit):
    # Convert 'from_unit' to meters (base unit)
    value_in_meters = value * conversion_rates['length'][from_unit]
    # Convert from meters to 'to_unit'
    converted_value = value_in_meters / conversion_rates['length'][to_unit]
    return converted_value

def convert_weight(value, from_unit, to_unit):
    # Convert 'from_unit' to kilograms (base unit)
    value_in_kilograms = value * conversion_rates['weight'][from_unit]
    # Convert from kilograms to 'to_unit'
    converted_value = value_in_kilograms / conversion_rates['weight'][to_unit]
    return converted_value

def convert_temperature(value, from_unit, to_unit):
    # Convert 'from_unit' to Kelvin (base unit)
    value_in_kelvin = conversion_rates['temperature'][from_unit]['to_kelvin'](value)
    # Convert from Kelvin to 'to_unit'
    converted_value = conversion_rates['temperature'][to_unit]['from_kelvin'](value_in_kelvin)
    return converted_value

print("Conversion functions defined.")
```

```python
def run_converter():
    while True:
        choice = display_main_menu()

        if choice == 0:
            print("Exiting Unit Converter. Goodbye!")
            break

        category_names = list(conversion_rates.keys())
        selected_category = category_names[choice - 1]

        try:
            value = float(input(f"Enter the value to convert in {selected_category} units: "))
        except ValueError:
            print("Invalid value. Please enter a number.")
            continue

        if selected_category == 'temperature':
            available_units = list(conversion_rates['temperature'].keys())
            from_unit, to_unit = get_unit_input(selected_category, available_units)
            result = convert_temperature(value, from_unit, to_unit)
        else:
            available_units = list(conversion_rates[selected_category].keys())
            from_unit, to_unit = get_unit_input(selected_category, available_units)
            if selected_category == 'length':
                result = convert_length(value, from_unit, to_unit)
            elif selected_category == 'weight':
                result = convert_weight(value, from_unit, to_unit)
            else:
                print("Unknown category.") # Should not happen with valid choices
                continue

        print(f"\n{value} {from_unit} is equal to {result:.4f} {to_unit}.")

# To run the converter, uncomment the line below:
# run_converter()
```
