### 🗺️ Smart City Explorer



```python
# Import the Google GenAI SDK and Colab secrets helper
from google import genai
from google.colab import userdata

# Initialize the client
try:
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
    client = genai.Client(api_key=GOOGLE_API_KEY)
except Exception as e:
    print("Error retrieving API Key. Please ensure GOOGLE_API_KEY is configured in your secrets: ", e)
```

```python
def get_mock_recommendations(city_name):
    """Provides structured mock recommendations as a fallback."""
    data = {
        "delhi": """🏛️ Historical
• Red Fort
• Qutub Minar
• Humayun's Tomb

🍴 Food
• Karim's (Chandni Chowk)
• Indian Accent
• Bukhara

🎨 Activities
• National Crafts Museum
• Lodhi Gardens
• Dilli Haat Market""",
        "new york": """🏛️ Historical
• Statue of Liberty
• Empire State Building
• Grand Central Terminal

🍴 Food
• Katz's Delicatessen
• Joe's Pizza
• Eleven Madison Park

🎨 Activities
• Metropolitan Museum of Art
• Central Park
• Chelsea Market"""
    }
    return data.get(city_name.lower(), f"""🏛️ Historical
• Local Historical Site 1
• Local Historical Site 2

🍴 Food
• Popular Local Diner
• Traditional Eatery

🎨 Activities
• City Central Park
• Local Art Gallery""")

def explore_city(city_name):
    prompt = f"""
    You are a Smart City Explorer guide.
    Provide recommendations for the city: {city_name}.

    Format the output exactly like this:
    🏛️ Historical
    • [Historical site 1]
    • [Historical site 2]
    • [Historical site 3]

    🍴 Food
    • [Popular local restaurant/food area 1]
    • [Popular local restaurant/food area 2]
    • [Popular local restaurant/food area 3]

    🎨 Activities
    • [Activity/Museum/Park 1]
    • [Activity/Museum/Park 2]
    • [Activity/Museum/Park 3]
    """

    try:
        response = client.models.generate_content(
            model='gemini-3.8-flash',
            contents=prompt
        )
        return response.text
    except Exception as api_err:
        # Fallback to mock data if API fails
        print(f"(API call unavailable. Using offline Explorer database...)\n")
        return get_mock_recommendations(city_name)

# User Input
city = "Delhi"
print(f"Exploring: {city}\n")

result = explore_city(city)
print(result)
```
