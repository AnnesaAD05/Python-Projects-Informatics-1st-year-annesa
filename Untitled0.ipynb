```python
import random

# --- Game Parameters ---
SECRET_NUMBER_MIN = 1
SECRET_NUMBER_MAX = 100
INITIAL_LIVES = 7

# Generate a random secret number
secret_number = random.randint(SECRET_NUMBER_MIN, SECRET_NUMBER_MAX)
lives = INITIAL_LIVES
guess = None
hints_given = [] # To keep track of hints already given

print(f"I'm thinking of a number between {SECRET_NUMBER_MIN} and {SECRET_NUMBER_MAX}.")
print(f"You have {lives} lives to guess it!")
```

```python
def give_hint():
    global hints_given
    available_hints = []

    # Hint 1: Even or Odd
    if 'parity' not in hints_given:
        parity_hint = 'even' if secret_number % 2 == 0 else 'odd'
        available_hints.append(f"The number is {parity_hint}.")

    # Hint 2: Multiple of 3 or 5 (if applicable)
    if 'multiple' not in hints_given:
        if secret_number % 3 == 0:
            available_hints.append("The number is a multiple of 3.")
        elif secret_number % 5 == 0:
            available_hints.append("The number is a multiple of 5.")

    # Choose a random hint from available ones
    if available_hints:
        hint = random.choice(available_hints)
        print(f"Hint: {hint}")
        if "even" in hint or "odd" in hint: # Check for parity hint specifically
            hints_given.append('parity')
        elif "multiple" in hint:
            hints_given.append('multiple')
    else:
        print("No new hints to give.")


while guess != secret_number and lives > 0:
    try:
        guess_str = input(f"You have {lives} lives left. Enter your guess: ")
        guess = int(guess_str)

        if not (SECRET_NUMBER_MIN <= guess <= SECRET_NUMBER_MAX):
            print(f"Please guess a number between {SECRET_NUMBER_MIN} and {SECRET_NUMBER_MAX}.")
            continue

        if guess < secret_number:
            print("Higher!")
            lives -= 1
        elif guess > secret_number:
            print("Lower!")
            lives -= 1
        else:
            print(f"Congratulations! You guessed the number {secret_number} in {INITIAL_LIVES - lives} tries!")
            break

        # Give a hint if lives are low and a hint hasn't been given yet
        if lives == INITIAL_LIVES // 2 and 'parity' not in hints_given:
            print("You're losing lives, here's a hint!")
            give_hint()
        elif lives == INITIAL_LIVES // 4 and 'multiple' not in hints_given:
            print("Another hint for you!")
            give_hint()

    except ValueError:
        print("Invalid input. Please enter a whole number.")

if lives == 0:
    print(f"Game Over! You ran out of lives. The secret number was {secret_number}.")
```
