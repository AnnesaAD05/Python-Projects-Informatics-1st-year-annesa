```python
import time
import sys

def pomodoro_timer(focus_minutes, break_minutes):
    """
    Runs a Pomodoro timer with specified focus and break times.
    """
    def countdown(total_seconds, phase_name):
        while total_seconds > 0:
            mins, secs = divmod(total_seconds, 60)
            timer = f'{phase_name}: {mins:02d}:{secs:02d}'
            sys.stdout.write('\r' + timer)
            sys.stdout.flush()
            time.sleep(1)
            total_seconds -= 1
        sys.stdout.write('\r' + ' ' * len(timer) + '\r')
        sys.stdout.flush()

    print(f"Starting Pomodoro Timer: {focus_minutes} min focus, {break_minutes} min break.\n")

    while True:
        print("\n--- FOCUS TIME ---")
        countdown(focus_minutes * 60, "Focus")
        print("Time for a break!\a")
        time.sleep(1)

        print("\n--- BREAK TIME ---")
        countdown(break_minutes * 60, "Break")
        print("Break is over! Back to work!\a")
        time.sleep(1)

```

```python
FOCUS_DURATION = 25
BREAK_DURATION = 5

# Start the Pomodoro timer
pomodoro_timer(FOCUS_DURATION, BREAK_DURATION)
```
